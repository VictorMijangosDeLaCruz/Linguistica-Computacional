#### Ejercicio: Extender la función de preoprocesamiento y eliminar stopwords

```python
from nltk.corpus import stopwords

def preprocess_corpus(corpus: list) -> list:
    lower_corpus = list(map(str.lower, corpus))
    result = list(filter(lambda word: word not in stopwords.words("spanish"), lower_corpus))
    return result
```

```python
corpus = preprocess_corpus(corpus)
```

```python
tokens = len(corpus) 
tipos = len(set(corpus))
```

```python
print_corpus_stats(tipos, tokens)
```

## Concordancias

Las concordancias o KWIC (Key Word In Context) nos permiten hacer búsquedas dentro de un corpus y obtener los contextos de aparición de las palabras. Las palabras se nos presentan alineadas en una ventana de un tamaño determinado. Con la paquetería de NLTK se pueden hacer estas búsquedas de manera fácil.

```python
# Indexar el corpus para hacer búsquedas
index = nltk.text.ConcordanceIndex(corpus)
```

```python
# Palabra de búsqueda
word = input("palabra a buscar>> ")
# Realizar e imprimir las concordancias
index.print_concordance(word, width=100, lines=10)
```

Los corpus paralelos son una herramienta primordial para tareas del PLN como la traducción automática. Aquí mostramos una paquetería de corpus paralelos en lenguas mexicanas. En primer lugar instalamos dicha paquetería.

Esta paquetería cuenta con corpus paralelos español-náhuatl y español-otomí.

```python
import elotl.corpus

#Cada elemento de esta lista corresponde a una oración paralela del corpus más información extra del par de oraciones
corpus_na = elotl.corpus.load('axolotl')  #español-náhuatl 
corpus_ot = elotl.corpus.load('tsunkua')  #español-otomi 
```

```python
table = Table(title="[bright_magenta]corpus data (Nahuatl)", show_lines=True)
table.add_column("l1")
table.add_column("l2")
table.add_column("variant")
table.add_column("source")
for row in corpus_na[:3]:
    table.add_row(*row)
rprint(table)
```

```python
table = Table(title="[bright_magenta]corpus data (Otomi)", show_lines=True)
table.add_column("l1")
table.add_column("l2")
table.add_column("variant")
table.add_column("source")
table.add_column("pdf file")
table.add_column("doc id")
for row in corpus_ot[:3]:
    table.add_row(*row)
rprint(table)
```

En el corpus paralelo, cada oración en una lengua se encuentra alineada con su correspondiente oración en la otra lengua.

```python
from rich.table import Table

def render_table(corpus: list, l1: str, l2: str, max_rows: int=5) -> None:
    table = Table(title=f"{l1.title()}-{l2.title()}", show_lines=True)
    table.add_column(f"[yellow]{l1} (l1)")
    table.add_column(f"[green]{l2} (l2)")
    for sent in corpus[:max_rows]:
        table.add_row(sent[0], sent[1])
    rprint(table)
```

```python
render_table(corpus_na, "spanish", "nahuatl")
```

```python
render_table(corpus_ot, "spanish", "otomi")
```
